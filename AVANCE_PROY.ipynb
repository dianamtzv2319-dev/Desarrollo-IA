```python
# Importamos la biblioteca NLTK y las herramientas para el Chat
import nltk
from nltk.chat.util import Chat, reflections

# Definición de pares: (Expresión regular de entrada, [Lista de posibles respuestas])
# El (.*) funciona como un comodín para que el usuario pueda escribir palabras antes o después de la palabra clave.
pares_adopcion = [
    # 1. Intención: Saludos
    (r'(.*)(hola|buenas|buenos dias|buenas tardes)(.*)',
     ['¡Hola! Bienvenido al asistente virtual de adopciones caninas. ¿En qué te puedo ayudar hoy?']),

    # 2. Intención: Requisitos de adopción
    (r'(.*)(requisito|requisitos|necesito para adoptar|como adopto)(.*)',
     ['Para adoptar necesitamos que seas mayor de edad. Los requisitos básicos son: copia de INE, comprobante de domicilio y llenar nuestro formulario de adopción. ¿Te gustaría saber el número de contacto?']),

    # 3. Intención: Información de esterilización
    (r'(.*)(operados|esterilizad[oa]s|esterilizacion)(.*)',
     ['Cuidamos mucho la salud de nuestros rescatados. Los perritos adultos se entregan ya esterilizados. En el caso de los cachorros, se entregan desparasitados y con una estricta promesa de esterilización.']),

    # 4. Intención: Contacto específico (Ejemplo: preguntar por Miel o pedir número)
    (r'(.*)(contacto|telefono|numero|miel|informes)(.*)',
     ['¡Claro que sí! Para atención directa, iniciar tu proceso o pedir informes específicos de nuestros perritos (como Miel), por favor comunícate al número: 729 8799346.']),

    # 5. Intención: Despedida y agradecimiento
    (r'(.*)(gracias|adios|bye|hasta luego)(.*)',
     ['¡Gracias a ti por tu interés en cambiarle la vida a un perrito! Que tengas un excelente día.']),

    # 6. Intención: Manejo de errores (Respuesta por defecto si no entiende)
    (r'(.*)',
     ['Lo siento, no entendí muy bien tu mensaje. ¿Podrías preguntarme sobre los requisitos de adopción, si están operados, o pedir nuestro número de contacto?'])
]

def iniciar_chatbot():
    print("=====================================================")
    print("INICIANDO CHATBOT DE ADOPCIONES CANINAS ")
    print("Escribe 'salir' para terminar la conversación.")
    print("=====================================================\n")

    # Instanciamos el Chat de NLTK pasándole nuestras reglas (pares)
    chatbot_rescate = Chat(pares_adopcion, reflections)

    # Bucle infinito para mantener la conversación activa hasta que el usuario escriba 'salir'
    while True:
        try:
            entrada = input("Tú: ")

            # Condición de salida
            if entrada.lower() == 'salir':
                print("Chatbot: ¡Hasta luego! Gracias por apoyar el rescate animal.")
                break

            # Generar y mostrar respuesta
            respuesta = chatbot_rescate.respond(entrada.lower())
            print("Chatbot:", respuesta)

        except EOFError:
            break

# Este bloque asegura que el bot se inicie automáticamente al ejecutar la celda
if __name__ == "__main__":
    iniciar_chatbot()
```
